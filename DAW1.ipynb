```python
emp_details = {
    'EmpID': [101,102,103,104,105],
    'City': ['Hyderabad','Chennai','Bangalore',None,'Hyderabad'],
    'Experience': [None,4,3,5,2],
    'Age': [25,30,None,35,25]
}
data = {
    "EmpID" : [101,102,103,104,105,106,107,108],
    "Name" : ["Ravi","Priya","Arun", "Sneha", "Kiran", "Anjali", "Vikram", "Meena"],
    "Age" : [22, 24, None, 25, None, 26, 20, 27],
    "City" : ["Hyderabad", "Chennai", "Bangalore", "Hyderabad", "Mumbai", None, "Bangalore", "Mumbai"],
    "Salary" : [30000, 40000, 25000, 50000, 35000, 45000, None, 55000]
}
df1= pd.DataFrame(data)
df = pd.DataFrame(emp_details)
print(df)
df.isna()
df.isna().sum()
df.fillna({'City' : 'Unknown', 'Experience': 0, 'Age': df['Age'].mean()}, inplace=True)
#df.rename(columns={'EmpID':'Employee_ID'}, inplace=True)
df['Salary']= [35000,23000,25000,46000,78000]
df['Age'] = df['Age'] + 1
df['Bonus'] = df['Salary'] / 10
df.drop('Bonus', axis= 1, inplace=True) #to drop a SINGLE column
df.drop(0,axis=0,inplace=True) #to drop a single row; [0,1] to drop MULTIPLE
#df.drop(columns=['Salary'],inplace=True) #to drop MULTIPLE columns
#df.loc[len(df)]= [109, 'Seattle', 7, 40, 57000]
print(df.groupby('City')['Age'].mean()) #groups by city w the mean age.
print(df.groupby('City')['Age'].agg(['mean', 'max','min'])) #used .agg to perform multiple aggregations on a SINGLE coulmn
print(df.groupby('City').agg({'Age' : ['mean','max'], 'Salary' : ['sum', 'count']})) #can use .agg to perform multiple aggregations on MULTIPLE columns
print(df)
merged = pd.merge( df, df1, on = 'EmpID', how = 'left') #used to merge two DF's with left, right, inner and outer forms
print(merged)

df.dropna(axis=0)
df.dropna(axis=1)
df.dropna(how = 'all')
df.dropna(inplace=True)
```

```python
data = {
  "VisitID": [501,502,503,504,505,506,507,508,509,510,510,511],
  "Patient": ["Rahul","Anita","Suresh","Priya","Kiran","Anjali","Ravi","Meena","Arun","Priya","Priya","Kiran"],
  "Department": ["Cardiology","Neurology","Orthopedic","General","Dental","Cardiology","Neurology","General","Dental","Cardiology","Cardiology","Dental"],
  "BillAmount": [5000,7500,4500,2000,1500,6000,7000,2500,1800,6000,6000,1500]}
df_patient = pd.DataFrame(data)

print(df_patient)
duplicates = df_patient[df_patient.duplicated()] #1
print(duplicates)
duplicate_count = df_patient.duplicated().sum() #2
print(f"Number of duplicate records: {duplicate_count}")

all_duplicates = df_patient[df_patient.duplicated(keep=False)] #3
print(all_duplicates)

duplicate_bill_total = df_patient[df_patient.duplicated()]['BillAmount'].sum() #4
print(f"Total BillAmount of duplicates: {duplicate_bill_total}")

visit_id_duplicates = df_patient[df_patient.duplicated(subset=['VisitID'], keep=False)] #5
print(visit_id_duplicates)

returning_patients = df_patient[df_patient.duplicated(subset=['Patient'], keep=False)] #6
print(returning_patients)

df_no_full_dupes = df_patient.drop_duplicates() #7
df_unique_visits = df_patient.drop_duplicates(subset=['VisitID']) #8
df_unique_patient_dept = df_patient.drop_duplicates(subset=['Patient', 'Department']) #9
df_cleaned = df_patient.drop_duplicates(subset=['VisitID']).copy() #10
df_cleaned.reset_index(drop=True, inplace=True)

print("Final Cleaned Dataset:")
print(df_cleaned)
#df["Date"]= pd.to_datetime(df["Date"]) #to convert the column to Date and Time data type
```
